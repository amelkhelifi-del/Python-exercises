```python
import numpy as np

S = np.array([
    
    [10, 4],
    [2, 1],
    [7, 3],
    [1, 2]
])

theta = np.array([0.3, 0.7])
bias = 0.5

# Calculate z
z = S @ theta + bias
print("z =", z)

# Sigmoid
sigmoid = 1 / (1 + np.exp(-z))
print("Probabilities =", sigmoid)

# Spam prediction
spam = sigmoid >= 0.5
print("Spam =", spam)


```

```python


import pandas as pd

df = pd.read_csv("logins.csv")



ru_suspicious = df[
(df["country"] == "RU") &
(df["failed_attempts"] >= 5)
]

print("Question 1:")
print("RU suspicious logins =", len(ru_suspicious))




total_failed = (
df.groupby("user")["failed_attempts"]
.sum()
.sort_values(ascending=False)
)

print("\nQuestion 2:")
print(total_failed)

print("User with highest failed attempts =", total_failed.index[0])
print("Total =", total_failed.iloc[0])



df["long_session"] = (df["duration_min"] > 30).astype(int)

print("\nQuestion 3:")
print(df[["duration_min", "long_session"]].head())


print("\nFinal DataFrame:")
print(df.head())
```
